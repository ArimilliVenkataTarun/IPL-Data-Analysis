```python
matches['toss_winner'].value_counts().plot(kind='bar')

plt.title("Toss Winners")

plt.show()
```

```python
matches['toss_decision'].value_counts().plot(kind='pie', autopct='%1.1f%%')

plt.title("Toss Decision Analysis")

plt.show()
```

```python
matches['player_of_match'].value_counts().head(10).plot(kind='bar')

plt.title("Top Player of the Match Winners")

plt.show()
```

```python
batsman_runs = deliveries.groupby('batter')['batsman_runs'].sum()

batsman_runs.sort_values(ascending=False).head(10)
```

```python
wickets = deliveries[deliveries['is_wicket'] == 1]

wickets['bowler'].value_counts().head(10).plot(kind='bar')

plt.title("Top Wicket Takers")

plt.show()
```

```python
matches['venue'].value_counts().head(10).plot(kind='bar')

plt.title("Most Used IPL Venues")

plt.show()
```

```python
matches['season'].value_counts().sort_index().plot(kind='line')

plt.title("Matches Per Season")

plt.show()
```
